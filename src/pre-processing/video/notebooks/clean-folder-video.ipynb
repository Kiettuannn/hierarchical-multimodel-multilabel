```python
import os
import pandas as pd

csv_path = r'/data/label-origin.csv.csv'
video_dir = r'/data/videos'

# Load the CSV file
df = pd.read_csv(csv_path)

# Extract valid video filenames into a set for fast lookup
valid_videos = set(df['filename'].dropna().tolist())

print(f"Number of videos in CSV: {len(valid_videos)}")

deleted_count = 0
kept_count = 0

# Check if directory exists
if os.path.exists(video_dir):
    for filename in os.listdir(video_dir):
        file_path = os.path.join(video_dir, filename)
        
        # Only process files
        if os.path.isfile(file_path):
            if filename not in valid_videos:
                try:
                    os.remove(file_path)
                    deleted_count += 1
                except Exception as e:
                    print(f"Error deleting {filename}: {e}")
            else:
                kept_count += 1
else:
    print(f"Directory does not exist: {video_dir}")

print(f"Finished processing.")
print(f"Videos kept: {kept_count}")
print(f"Videos deleted: {deleted_count}")
```
