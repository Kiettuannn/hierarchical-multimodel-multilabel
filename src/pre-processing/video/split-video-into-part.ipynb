```python
import os
import shutil
from tqdm import tqdm
import math

VIDEO_DIR = "../../../data/videos"
OUTPUT_BASE_DIR = "../../../data/video_parts"
BATCH_SIZE = 500

os.makedirs(OUTPUT_BASE_DIR, exist_ok=True)

# Retrieve all video file paths
video_files = [f for f in os.listdir(VIDEO_DIR) if f.endswith(('.mp4', '.avi'))]
video_files.sort()
total_videos = len(video_files)
print(f"Total videos found: {total_videos}")

num_parts = math.ceil(total_videos / BATCH_SIZE)
print(f"Splitting into {num_parts} parts (max {BATCH_SIZE} videos per part)")

def create_part(part_index):
    start_idx = part_index * BATCH_SIZE
    end_idx = min((part_index + 1) * BATCH_SIZE, total_videos)
    
    if start_idx >= total_videos:
        print(f"Part {part_index + 1} is empty, skipping.")
        return
        
    part_dir = os.path.join(OUTPUT_BASE_DIR, f"part_{part_index + 1:02d}")
    os.makedirs(part_dir, exist_ok=True)
    
    batch = video_files[start_idx:end_idx]
    
    # Using shutil.move to avoid duplicating data and filling up disk space.
    # Change to shutil.copy if you want to keep the original folder intact.
    for filename in tqdm(batch, desc=f"Moving to part_{part_index + 1:02d}"):
        src_path = os.path.join(VIDEO_DIR, filename)
        dst_path = os.path.join(part_dir, filename)
        shutil.move(src_path, dst_path)

print("Setup complete. Run subsequent cells to split videos.")
```
